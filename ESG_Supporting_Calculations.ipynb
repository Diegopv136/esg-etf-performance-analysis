```python
import pandas as pd
import yfinance as yf
```

```python
# 1. U.S. RISK-FREE RATE (^IRX on Yahoo Finance, last 5 years)
us_tbill = yf.download("^IRX", period="5y", auto_adjust=True)["Close"]
us_rf = float(us_tbill.mean().iloc[0]) / 100
print(f"1. Average U.S. Risk-Free Rate: {us_rf*100:.2f}% (Decimal: {us_rf:.4f})")
```

```python
# 2. CANADIAN RISK-FREE RATE (tbill_all.csv, skipping the first 24 text rows)
ca_tbill = pd.read_csv("tbill_all.csv", skiprows=24)
ca_tbill_5y = ca_tbill[ca_tbill["date"] >= "2021-10-14"]  # Filter last 5 years
ca_rf = ca_tbill_5y["V80691303"].mean() / 100
print(f"2. Average Canadian Risk-Free Rate: {ca_rf*100:.2f}% (Decimal: {ca_rf:.4f})")
```

```python
# 3. U.S. INFLATION (CPIAUCSL.csv from FRED: Oct 2021 to Sep 2026)
us_cpi = pd.read_csv("CPIAUCSL.csv")
us_cpi_5y = us_cpi[us_cpi["observation_date"] >= "2021-10-01"]
us_cpi_start = us_cpi_5y["CPIAUCSL"].iloc[0]   # 276.550 (Oct 2021)
us_cpi_end = us_cpi_5y["CPIAUCSL"].iloc[-1]    # 336.298 (Sep 2026)
us_inf = (us_cpi_end / us_cpi_start) ** (1 / 5) - 1
print(f"3. Annualized U.S. Inflation({us_cpi_start} -> {us_cpi_end}): {us_inf*100:.2f}% (Decimal: {us_inf:.4f})")
```

```python
# 4. CANADIAN INFLATION (CPI_MONTHLY.csv, skipping the first 25 text rows)
ca_cpi = pd.read_csv("CPI_MONTHLY.csv", skiprows=25)
ca_cpi_5y = ca_cpi[ca_cpi["date"] >= "2021-10-01"]

# Column V41690973 (Not seasonally adjusted: 143.9 -> 169.8)
ca_inf_unadj = (ca_cpi_5y["V41690973"].iloc[-1] / ca_cpi_5y["V41690973"].iloc[0]) ** (1 / 5) - 1
# Column V41690914 (Seasonally adjusted: 143.9 -> 169.3)
ca_inf_sa = (ca_cpi_5y["V41690914"].iloc[-1] / ca_cpi_5y["V41690914"].iloc[0]) ** (1 / 5) - 1

print(f"4a. Canadian Inflation (Not Seasonally Adjusted) (143.9 -> 169.8): {ca_inf_unadj*100:.2f}% (Decimal: {ca_inf_unadj:.4f})")
print(f"4b. Canadian Inflation (Seasonally Adjusted) (143.9 -> 169.3): {ca_inf_sa*100:.2f}% (Decimal: {ca_inf_sa:.4f})")
```
